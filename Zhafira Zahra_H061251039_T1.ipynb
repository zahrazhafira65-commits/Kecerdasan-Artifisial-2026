# 5 Prompt Geofisika

## Struktur Prompt

- **R — Role**: Peran AI
- **T — Task**: Tugas yang harus dikerjakan
- **C — Context**: Latar belakang masalah

# Prompt 1 — Interpretasi Anomali Bouguer

```text
# R — Role
Anda berperan sebagai ahli geofisika eksplorasi yang membimbing mahasiswa S1 Geofisika dalam menganalisis dan menginterpretasikan data anomali Bouguer.

# T — Task
Analisis distribusi anomali Bouguer pada suatu wilayah untuk mengidentifikasi pola anomali positif dan negatif serta memberikan interpretasi awal mengenai variasi densitas material bawah permukaan.

# C — Context
Metode gravitasi memanfaatkan variasi medan gravitasi bumi untuk mengetahui perbedaan densitas material di bawah permukaan. Anomali Bouguer dapat menunjukkan adanya perubahan densitas yang berkaitan dengan kondisi geologi bawah permukaan. Analisis dilakukan untuk memahami hubungan antara pola anomali gravitasi dan kemungkinan struktur atau variasi litologi di wilayah penelitian.

# I — Input
Gunakan data yang diberikan berupa koordinat titik pengukuran, nilai gravitasi atau anomali Bouguer, serta informasi geologi regional jika tersedia. Jika tersedia peta kontur anomali, gunakan informasi tersebut untuk membantu menentukan pola spasial anomali.

# K — Constraints
- Jangan mengubah nilai asli data.
- Jelaskan jenis anomali yang digunakan sebelum melakukan interpretasi.
- Bedakan antara hasil pengamatan data dan interpretasi geologi.
- Jangan menentukan jenis batuan secara pasti hanya berdasarkan nilai anomali gravitasi.
- Jangan menyimpulkan keberadaan struktur geologi tertentu tanpa dukungan data.
- Jika informasi yang tersedia tidak cukup, jelaskan keterbatasannya.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset dan wilayah pengamatan.
2. Identifikasi zona anomali Bouguer relatif tinggi dan rendah.
3. Penjelasan pola distribusi anomali secara spasial.
4. Interpretasi kemungkinan variasi densitas bawah permukaan.
5. Kemungkinan hubungan anomali dengan kondisi geologi regional.
6. Interpretasi hasil dalam bentuk paragraf.
7. Keterbatasan analisis dan kesimpulan singkat.

# Q — Quality Criteria
Interpretasi harus sesuai dengan prinsip dasar metode gravitasi, menggunakan istilah geofisika yang tepat, membedakan fakta data dengan dugaan interpretasi, dan mudah dipahami oleh mahasiswa S1 Geofisika.

# V — Verification
Periksa apakah zona anomali tinggi dan rendah benar-benar sesuai dengan nilai data, apakah pola spasial pada peta sesuai dengan deskripsi, dan apakah interpretasi variasi densitas tidak melebihi informasi yang dapat didukung oleh dataset.
```


# Prompt 2 — Analisis Magnitudo dan Energi Gempa

```text
# R — Role
Anda berperan sebagai seismolog yang membimbing mahasiswa S1 Geofisika dalam menghitung dan memahami hubungan antara magnitudo gempa dan energi yang dilepaskan.

# T — Task
Gunakan data magnitudo gempa yang diberikan untuk menghitung perkiraan energi yang dilepaskan, membandingkan beberapa kejadian gempa, dan menjelaskan hubungan antara peningkatan magnitudo dengan perubahan energi.

# C — Context
Magnitudo merupakan ukuran kuantitatif yang berkaitan dengan ukuran atau energi gempa. Hubungan antara magnitudo dan energi bersifat nonlinier, sehingga kenaikan magnitudo yang relatif kecil dapat menghasilkan perbedaan energi yang besar. Analisis dilakukan sebagai latihan untuk memahami makna fisik magnitudo dalam seismologi.

# I — Input
Gunakan data berupa nilai magnitudo satu atau beberapa kejadian gempa. Jika tersedia, gunakan pula informasi waktu kejadian, lokasi, kedalaman, dan jenis magnitudo yang digunakan.

# K — Constraints
- Jangan mengubah nilai magnitudo yang diberikan.
- Gunakan persamaan hubungan magnitudo dan energi yang sesuai serta tuliskan persamaan tersebut.
- Jelaskan satuan energi yang digunakan.
- Jangan menyamakan magnitudo dengan intensitas gempa.
- Jika jenis magnitudo tidak diketahui, nyatakan asumsi yang digunakan.
- Hindari memberikan kesimpulan mengenai tingkat kerusakan hanya berdasarkan magnitudo.

# O — Output
Hasil harus mencakup:
1. Data magnitudo yang digunakan.
2. Persamaan untuk memperkirakan energi gempa.
3. Langkah perhitungan energi untuk setiap kejadian.
4. Perbandingan energi antarkejadian.
5. Penjelasan mengenai perubahan energi akibat kenaikan magnitudo.
6. Interpretasi hasil dalam konteks seismologi.
7. Kesimpulan singkat.

# Q — Quality Criteria
Perhitungan harus benar, satuan konsisten, persamaan dijelaskan dengan jelas, dan interpretasi mampu menjelaskan hubungan magnitudo dan energi secara sederhana tetapi ilmiah.

# V — Verification
Periksa kembali hasil perhitungan dengan memasukkan nilai magnitudo ke persamaan, pastikan satuan energi konsisten, dan periksa apakah perbandingan energi antar-gempa sesuai dengan sifat hubungan magnitudo-energi.
```


# Prompt 3 — Analisis Hubungan Suhu dan Kelembapan Udara

```text
# R — Role
Anda berperan sebagai ahli meteorologi dan klimatologi yang membimbing mahasiswa S1 Geofisika dalam menganalisis hubungan antara suhu udara dan kelembapan relatif.

# T — Task
Analisis pola perubahan suhu udara dan kelembapan relatif terhadap waktu serta hubungan statistik antara kedua parameter untuk mengidentifikasi pola atmosfer yang terdapat pada dataset.

# C — Context
Suhu udara dan kelembapan relatif merupakan parameter penting dalam kajian atmosfer. Perubahan suhu dapat memengaruhi kelembapan relatif karena kemampuan udara dalam menampung uap air berubah terhadap suhu. Hubungan kedua parameter dapat dianalisis menggunakan data pengamatan untuk mengetahui pola harian maupun pola pada periode tertentu.

# I — Input
Gunakan dataset yang berisi waktu pengamatan, suhu udara, dan kelembapan relatif. Jika tersedia, gunakan data tambahan seperti curah hujan, tekanan udara, dan kecepatan angin sebagai informasi pendukung.

# K — Constraints
- Jangan mengubah nilai asli dataset.
- Identifikasi dan jelaskan data kosong atau nilai yang tidak valid.
- Jangan menyimpulkan hubungan sebab-akibat hanya berdasarkan korelasi.
- Jelaskan satuan setiap parameter.
- Bedakan pola harian, musiman, dan kejadian khusus jika dapat diidentifikasi dari data.
- Jangan menganggap korelasi yang tinggi selalu menunjukkan hubungan sebab-akibat.

# O — Output
Hasil harus mencakup:
1. Deskripsi periode dan kelengkapan dataset.
2. Pola perubahan suhu terhadap waktu.
3. Pola perubahan kelembapan relatif terhadap waktu.
4. Hubungan antara suhu dan kelembapan relatif.
5. Nilai dan arah korelasi jika dapat dihitung.
6. Identifikasi pola atau nilai yang tidak biasa.
7. Interpretasi meteorologis dalam bentuk paragraf.
8. Kesimpulan singkat.

# Q — Quality Criteria
Analisis harus berdasarkan data aktual, menggunakan konsep meteorologi yang benar, menjelaskan hubungan suhu dan kelembapan secara logis, serta membedakan hasil statistik dengan interpretasi fisik.

# V — Verification
Periksa apakah pola yang dijelaskan sesuai dengan data, apakah nilai korelasi dihitung dengan benar, apakah nilai ekstrem memengaruhi hubungan secara berlebihan, dan apakah interpretasi meteorologis sesuai dengan pola yang ditemukan.
```


# Prompt 4 — Analisis DEM untuk Identifikasi Zona Lereng Berpotensi Longsor

```text
# R — Role
Anda berperan sebagai ahli geomorfologi dan Sistem Informasi Geografis yang membimbing mahasiswa S1 Geofisika dalam menganalisis kondisi topografi dan faktor spasial yang berkaitan dengan potensi longsor.

# T — Task
Analisis data DEM dan parameter pendukung untuk mengidentifikasi wilayah dengan karakteristik topografi yang dapat meningkatkan potensi terjadinya longsor.

# C — Context
Kemiringan lereng merupakan salah satu faktor yang berpengaruh terhadap kestabilan lereng. Namun, kondisi lereng tidak hanya dipengaruhi oleh kemiringan, tetapi juga dapat berkaitan dengan elevasi, litologi, penggunaan lahan, drainase, dan kondisi lingkungan. Oleh karena itu, analisis dilakukan dengan mempertimbangkan beberapa parameter jika datanya tersedia.

# I — Input
Gunakan data DEM, peta kemiringan lereng, elevasi, penggunaan lahan, litologi, dan parameter pendukung lainnya jika tersedia. Gunakan batas wilayah penelitian sebagai acuan analisis spasial.

# K — Constraints
- Jangan mengubah nilai elevasi atau data asli.
- Jelaskan metode dan klasifikasi kemiringan yang digunakan.
- Bedakan antara zona dengan lereng curam dan zona yang benar-benar dinyatakan berpotensi longsor.
- Jangan menentukan tingkat kerawanan hanya berdasarkan kemiringan lereng jika parameter lain tersedia.
- Jangan menyatakan suatu wilayah pasti mengalami longsor.
- Jelaskan keterbatasan data yang digunakan.

# O — Output
Hasil harus mencakup:
1. Deskripsi data DEM dan parameter pendukung.
2. Distribusi kelas kemiringan lereng.
3. Identifikasi wilayah dengan lereng rendah, sedang, dan tinggi.
4. Perbandingan kemiringan dengan penggunaan lahan dan litologi jika tersedia.
5. Identifikasi zona yang memiliki kombinasi faktor pendukung potensi longsor.
6. Interpretasi spasial dalam bentuk paragraf.
7. Keterbatasan analisis dan kesimpulan.

# Q — Quality Criteria
Analisis harus menggunakan konsep geomorfologi dan SIG yang benar, mempertimbangkan lebih dari satu faktor jika data tersedia, serta tidak memberikan klaim kerawanan yang melebihi kemampuan dataset.

# V — Verification
Periksa apakah kelas kemiringan sesuai dengan nilai DEM, apakah lokasi lereng curam sesuai dengan pola topografi, apakah overlay antarparameter dilakukan dengan benar, dan apakah zona yang diidentifikasi memiliki dasar data yang jelas.
```


# Prompt 5 — Analisis ENSO dan SOI terhadap Variabilitas Iklim

```text
# R — Role
Anda berperan sebagai ahli klimatologi dan geofisika atmosfer yang membimbing mahasiswa S1 Geofisika dalam menganalisis variabilitas iklim menggunakan indeks ENSO dan Southern Oscillation Index (SOI).

# T — Task
Analisis deret waktu ENSO dan SOI untuk mengidentifikasi perubahan fase ENSO, durasi kejadian, pola perubahan SOI, serta hubungan kedua indeks terhadap variabilitas iklim.

# C — Context
ENSO merupakan salah satu fenomena variabilitas iklim utama di wilayah tropis yang melibatkan interaksi antara laut dan atmosfer di Samudra Pasifik. SOI merupakan salah satu indikator perubahan tekanan atmosfer yang berkaitan dengan ENSO. Perubahan kedua indeks dapat digunakan untuk mempelajari karakteristik kejadian El Niño dan La Niña serta kaitannya dengan variabilitas iklim di wilayah tropis.

# I — Input
Gunakan data bulanan ENSO dan SOI pada periode pengamatan yang diberikan. Data dapat berupa tanggal, nilai indeks, dan informasi tambahan seperti curah hujan atau suhu jika tersedia.

# K — Constraints
- Jangan mengubah nilai asli indeks.
- Pertahankan urutan waktu dataset.
- Jelaskan kriteria yang digunakan untuk menentukan fase ENSO.
- Tentukan durasi kejadian berdasarkan data yang tersedia.
- Jangan menyatakan bahwa ENSO merupakan satu-satunya penyebab perubahan iklim atau cuaca lokal.
- Bedakan antara hubungan statistik dan hubungan fisik.
- Jika data tidak cukup untuk menjelaskan dampak terhadap Indonesia, nyatakan keterbatasannya.

# O — Output
Hasil harus mencakup:
1. Deskripsi periode dan kelengkapan dataset.
2. Pola perubahan ENSO dan SOI terhadap waktu.
3. Identifikasi periode El Niño, La Niña, dan kondisi netral berdasarkan kriteria yang digunakan.
4. Durasi setiap kejadian yang berhasil diidentifikasi.
5. Perbandingan pola ENSO dan SOI.
6. Analisis hubungan antara kedua indeks.
7. Pembahasan kemungkinan keterkaitannya dengan variabilitas iklim Indonesia jika data pendukung tersedia.
8. Keterbatasan analisis dan kesimpulan.

# Q — Quality Criteria
Analisis harus menggunakan konsep ENSO dan SOI secara tepat, mempertahankan urutan waktu, memberikan kriteria identifikasi yang jelas, serta membedakan hasil berdasarkan data dari interpretasi klimatologis.

# V — Verification
Periksa apakah fase ENSO yang diidentifikasi sesuai dengan nilai indeks dan kriteria yang digunakan, apakah durasi kejadian dihitung dengan benar, apakah pola SOI konsisten dengan perubahan ENSO, dan apakah kesimpulan didukung oleh data yang tersedia.
```
